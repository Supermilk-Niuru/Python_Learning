```python
# 将文件名以列表的形式存放，方便后续读取文件
files = []
for file in os.listdir('GSE162164_RAW'):
    if(file.endswith('.txt.gz')):
        files.append(file)
print(files)
pd.array(files)
```

```python
dfs = []
for file in files:
    # 转置：给的数据样本：特征在行，标签在列。
    df = (pd.read_csv("GSE162164_RAW/"+ file, sep = '\t')).T
    # 乘1.0是为了转化成浮点型数据
    dfs.append(df*1.0)
# 合并
dfs_con = pd.concat(dfs)
dfs_con
```

```python
marks = []
for sample in dfs_con.index:
    if sample.startswith("TB"):
        marks.append('TB+HIV')
    else:
        marks.append('HIV')
print(marks)
```

```python
print(marks.count("TB+HIV"))
print(marks.count("HIV"))
```

```python
TB = []
HIV = []
for sample in dfs_con.index:
    if sample.startswith("TB"):
        TB.append(sample)
    else:
        HIV.append(sample)
```

```python
TB
```

```python
HIV
```

```python
df_TB = dfs_con.loc[TB]
df_HIV = dfs_con.loc[HIV]
```

```python
df_TB
```

```python
df_HIV
```

```python
# 比较两组中每个特征的均值
mean_TB = df_TB.mean()
mean_HIV = df_HIV.mean()
```

```python
mean_TB
```

```python
mean_HIV
```

```python
# 均值进行差值比较，从大到小排列
mean_abs = np.abs(mean_TB-mean_HIV)
print(mean_abs.sort_values(ascending = False).head(100))
# 通过结果可以看出，仅比较均值远远不够
```
