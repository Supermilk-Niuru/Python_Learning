```python
# 将文件名以列表的形式存放，方便后续读取文件
files = []
for file in os.listdir('GSE162164_RAW'):
    if(file.endswith('.txt.gz')):
        files.append(file)
print(files)
pd.array(files)
```

```python
dfs = []
for file in files:
    # 转置：给的数据样本：特征在行，标签在列。
    df = (pd.read_csv("GSE162164_RAW/"+ file, sep = '\t')).T
    # 乘1.0是为了转化成浮点型数据
    dfs.append(df*1.0)
# 合并
dfs_con = pd.concat(dfs)
dfs_con
```

```python
marks = []
for sample in dfs_con.index:
    if sample.startswith("TB"):
        marks.append('TB')
    else:
        marks.append('HIV+TB')
print(marks)
```
