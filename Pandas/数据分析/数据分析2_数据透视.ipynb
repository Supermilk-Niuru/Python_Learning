```python
import pandas as pd
```

```python
df = pd.read_csv('Titanic.csv',index_col = 0)
df
```

```python
# index，columns 决定“怎么分组”，values 决定“算什么”，aggfunc 决定“怎么算”
df1 = df.pivot_table(values = '是否生还',index = '性别',columns = '船舱等级',aggfunc = 'mean') # 默认聚合函数aggfunc = 'mean' 也可以改成自己写的函数
df1
```

```python
df2 = df.pivot_table(values = '是否生还',index = '性别',aggfunc = 'mean')
df2
```

```python
# 多组数据透视
# 有的特征有很多数据，很难归类，需要使用cut/qcut 将这些数据离散化，也就是重新分组
# cut手动设置切分范围；qcut只需告诉它分组数量，自动切分

# 重置年龄列
age = pd.cut(df['年龄'],[-1,25,120])
age
```

```python
# 重置费用列
fare = pd.qcut(df['费用'],q = 2)
fare
```

```python
df3 = df.pivot_table(values = '是否生还',index = [age,'性别'],columns = ['船舱等级',fare],aggfunc = 'mean')
df3
```
