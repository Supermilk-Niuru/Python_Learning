```python
import pandas as pd
```

```python
# 导入Excel时 文件后缀名必须是.csv
# 文件内必须要有行标签(为文档第一列)，列标签(为文档第一行)
# 读取时使用pd.read_csv()
read1 = pd.read_csv('Test.csv')
print(read1)
# 注意两次输出的区别，由于0，1，2……并不是特征，而是行标签，所以使用index_col=0告诉它，第0列是行标签
read2 = pd.read_csv('Test.csv',index_col = 0)
print('\n')
print(read2)
```

```python
print(read1.shape)
print(read2.shape)
```

```python
# 注意观察二者输出的区别，values只管输出数据，不管标签
arr1 = read1.values
print(arr1)
arr2 = read2.values
print('\n')
print(arr2)
```
