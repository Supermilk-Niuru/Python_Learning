```python
import pandas as pd
```

```python
v = [[15,'male'],[20,'female'],[30,'male'],[40,'female']]
i = ['first','second','third','forth']
col = ['age','sex']
df = pd.DataFrame(v,columns = col,index = i)
print(df)
```

```python
# 显式方法提取二维对象的行
df1 = df.loc['first',:]
print(df1)
# 显式方法提取二维对象的列
df2 = df.loc[:,'age']
df3 = df['age'] #列标签可以这样简写，因为本身是字典里面的键
print(df2)
print(df3)
# 隐式方法和显式方法类似 注意是iloc()
```
