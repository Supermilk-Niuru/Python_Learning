```python
import pandas as pd
```

```python
# 索引分为显式索引 loc() 和隐式索引 iloc()
# 显式索引就是使用你提供给的pandas对象
# 隐式索引就是使用数组本身自带的从0开始

v = [100,90,89,77]
i = ['first','second','third','forth']
sr = pd.Series(v,index=i)
print(sr)
print(sr.loc['first']) # 显式索引
print(sr.iloc[0])      # 隐式索引
```

```python
# 切片
sr1 = sr['first':'third'] # index是标签,不是位置。pandas里面左右都包含
sr2 = sr[0:2] # 位置切片遵循python里面的原则，左闭右开
print(sr1)
print('\n') #换行符
print(sr2)
```
