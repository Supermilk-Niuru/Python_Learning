```python
import pandas as pd
```

```python
v1 = [10,'male','001']
i = ['age','sex','No']
sr1 = pd.Series(v1,index = i)
v2 = [20,'female','002']
sr2 = pd.Series(v2,index = i)
v3 = [30,'male','003']
sr3 = pd.Series(v3,index = i)
v4 = [40,'female','004']
sr4 = pd.Series(v4,index = i)
print(sr1,'\n',sr2,'\n',sr3,'\n',sr4)
```

```python
df = pd.DataFrame({'first':sr1,'second':sr2,'third':sr3,'forth':sr4})
print(df)
```

```python
# 转置
df_T = df.T
print(df_T)
```

```python
# 上下翻转
df1 = df_T.iloc[::-1,::] # 采样行间隔是-1 就是倒着采样
print(df1)
# 左右翻转
df2 = df_T.iloc[::,::-1]
print(df2)
```

```python
# 重塑/合并
arr1 = [100,200,300,400]
df_T['salary'] = arr1
print(df_T)
```

```python
# 拆分
arr2 = df_T['salary']
print(arr2)
```

```python
# 一维对象之间的合并
sr5 = pd.concat([sr1,sr2])
print(sr5)
sr5.index
```
