```python
import pandas as pd
```

```python
# 字典创建法 ： 每次给的是一个特征所有样本的数据，然后再拼接
# 按列思考

v1 = [12,13,4,5] # age
i = [1,2,3,4]
sr1 = pd.Series(v1,index = i)
print(sr1)

v2 = ['male','famale','male','female']
sr2 = pd.Series(v2,index = i)
print(sr2)

df = pd.DataFrame({'age':sr1,'sex':sr2})
print(df)
```

```python
# 二维列表创建法 ： 直接给的是一个样本的所有数据
# 按行思考
v = [[12,'male'],[13,'female'],[4,'male'],[5,'female']]
i = [1,2,3,4]
col = ['age','sex']
df = pd.DataFrame(v,columns = col,index = i)
print(df)
```
