```python
import pandas as pd
```

```python
# 创建一维对象，使用列表创建
# 也可以使用数组，张量创建
v = [1,50,20,39]      # value
k = ['a','b','c','d'] # key
sr = pd.Series(v,index = k)
print(sr)
```

```python
# 查看属性
print(sr.values)
print(sr.index)
```

```python
print(sr.keys) 
print(sr.keys())
```

| 写法          | 类型   | 是否执行   |
| ----------- | ---- | ------ |
| `sr.index`  | 属性   | 直接返回   |
| `sr.values` | 属性   | 直接返回   |
| `sr.keys`   | 方法对象 | 不执行    |
| `sr.keys()` | 方法调用 | 执行返回结果 |
