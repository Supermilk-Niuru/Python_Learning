```python
import pandas as pd
```

```python
# 使用isnull()发现缺失值 以二维对象为例
v = [[12,'male'],[None,None],[13,'None'],[13,'male']]
i = ['first','second','third','forth']
col = ['age','sex']
df = pd.DataFrame(v,columns = col,index = i)
print(df)
```

```python
# 发现缺失值 缺失为True 不缺失为False
isnull = pd.isnull(df)
print(isnull)
```

```python
# 剔除缺失值 df.dropna()
# 对于一维对象比较方便，二维对象要么剔除含有缺失值的一整行，要么是一整列
# 一般行是个体，列是特征
drop_row = df.dropna()
print(drop_row)
drop_col = df.dropna(axis = 1)
print('\n')
print(drop_col)
```

| 参数       | 写法          | 判断依据        | 含义              | 例子                        |
| -------- | ----------- | ----------- | --------------- | ------------------------- |
| `how`    | `how='any'` | 是否存在缺失值     | **有一个缺失值就删除**   | 一行 `[1, 2, NaN]` → 删除     |
| `how`    | `how='all'` | 是否全部是缺失值    | **全部缺失才删除**     | 一行 `[NaN, NaN, NaN]` → 删除 |
| `thresh` | `thresh=n`  | **非缺失值的数量** | 至少有 `n` 个有效值才保留 | `thresh=3` → 至少3个非NaN     |

**如果是5个缺失值及以上就不保留，`thresh=df.shape[1]-5` df.shape[1]自动计算一行有多少个特征**
