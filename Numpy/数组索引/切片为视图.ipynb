```python
import numpy as np
```

```python
# 创建一个数组
arr1 = np.arange(10)
print(arr1)
```

```python
# 视图（view）就是：创建了一个新的数组变量，但是它和原数组共享同一块内存数据。 没有创建新的变量
# 原数组改了，视图也会跟着被修改
cut1 = arr1[2:4]
print(cut1)
cut1[0] = 1000
print(cut1)
print(arr1) # 会发现原数组也跟着修改
```

```python
# 创建一个矩阵
arr2 = np.arange(0,20).reshape(4,5)
print(arr2)
cut2 = arr2[0:2,0:2]
print(cut2)
cut2[0][0] = 100
print(cut2)
print(arr2) # 会发现原矩阵也会跟着改
```
