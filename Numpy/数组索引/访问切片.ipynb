```python
import numpy as np
```

```python
# 创建一个向量
arr1 = np.arange(40,50)
print(arr1)
# 创建一个矩阵
arr2 = np.arange(20,40).reshape(4,5)
print(arr2)
```

```python
# 向量的切片
print(arr1[1:4]) # 从第1个索引切到第4个索引之前
print(arr1[1:]) # 从第1个索引切到最后一个索引
print(arr1[:5]) # 从最开始的索引切到第5个索引之前
# 采样
print(arr1[::2]) # 每隔2个元素采样一次
print(arr1[1:-1:2]) # 从第1个索引开始切到倒数第1个索引之前，然后每隔2个元素采样一次
```

```python
# 矩阵的切片
print(arr2[1:3,2:3]) # 切行索引1到行索引3之前，列索引2到列索引3之前
print(arr2[::2,::3]) # 行隔2个索引采样一次，列隔三个索引采样一次
```

```python
# 行矩阵的提取
print(arr2[2,:]) #提取第2行，列都要
print(arr2[0:2,:]) #提取第0-1行，列都要
```

```python
# 列矩阵的提取
print(arr2[:,1]) # 会发现得到的不是矩阵，而是一个向量
print(arr2[:,1:3]) # 行都要，提取第1，2列
```

```python
# 列矩阵单列提取得到矩阵
print(arr2[:,1].reshape(-1,1)) # （-1，1）中 -1为自动计算多少行，1表示1列。使用塑形函数变成矩阵
# 还有另外一个处理办法，就是先将向量变成矩阵，然后进行转置操作 .T ->这个操作仅针对矩阵，向量无法进行这个操作
```
