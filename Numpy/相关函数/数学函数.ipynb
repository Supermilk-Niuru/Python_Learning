```python
# 聚合函数
arr3 = np.arange(10).reshape(2,5)
print(arr3)
print(arr3.shape)

# 主要有求和，求积，求最大最小，求平均，求标准差，求方差 sum() prod() max() min() mean() std() var()
# 大型数组可能出现缺失值，在函数名前加nan，来忽略缺失值 如：nansum()

print(np.sum(arr3,axis = 0)) # 列不动，逐行求和
print(np.sum(arr3,axis = 1)) # 行不动，逐列求和
print(np.sum(arr3)) # 整体求和
```
