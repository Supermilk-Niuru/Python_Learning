```python
import numpy as np
```

```python
# 创建向量
arr1 = np.arange(10)
print(arr1)
arr2 = np.arange(10,20)
print(arr2)
# 创建矩阵
arr3 = np.arange(20).reshape(4,-1) 
print(arr3)
arr4 = np.arange(20,40).reshape(4,-1)
print(arr4)
```

```python
# 向量的拼接
arr5 = np.concatenate([arr1,arr2])
print(arr5)
# 矩阵的拼接
arr6 = np.concatenate([arr3,arr4]) # 本处省去axis = 0，默认 axis = 0 就是按行拼接 
print(arr6)
arr7 = np.concatenate([arr3,arr4],axis = 1) # axis = 1 是按列拼接
print(arr7)
```

```python
# 向量分裂
arr8,arr9,arr10 = np.split(arr1,[2,8]) # 索引2，8位置截断
print(arr8,arr9,arr10)
```

```python
# 矩阵分裂
arr11,arr12,arr13 = np.split(arr6,[2,5]) # 默认axis=0 按行分裂
print(arr11,'\n\n',arr12,'\n\n',arr13)
arr14,arr15,arr16 = np.split(arr3,[2,4],axis = 1) # 按列分裂
print(arr14,'\n\n',arr15,'\n\n',arr16,'\n\n')
```
