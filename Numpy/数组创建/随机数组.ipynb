```python
import numpy as np
```

```python
# 浮点型随机数组
arr1 = np.random.random((1,10)) # 只可生成0-1之间的随机数 如果要生成其他的 加一个线性变换即可
print(arr1)
```

```python
# 整数型随机数组
arr2 = np.random.randint(1,100,(1,10)) # （1，10）用来规定矩阵的形状 （low，high，形状）生成随机数不包括high
print(arr2)
```

```python
# 标准正态分布
arr3 = np.random.normal(2,1,(1,10)) # (均值，标准差，形状)
print(arr3)
```
